```python
## Berechnung der Rezensionslänge

import pandas as pd

# Datei laden
df = pd.read_excel("Bewertungen mit BERT neu.xlsm")

# Anzahl Wörter berechnen
df["review_length"] = (
    df["review_text_clean"]
    .fillna("")
    .astype(str)
    .str.split()
    .str.len()
)

# Neue Datei speichern
df.to_excel("Bewertungen mit BERT neu_mit_review_length.xlsx", index=False)

print("Fertig!")
```

```python
import pandas as pd
import pyreadstat

# Excel laden
df = pd.read_excel("Bewertungen mit BERT neu.xlsm")

# Anzahl Wörter berechnen
df["review_length"] = (
    df["review_text_clean"]
    .fillna("")
    .astype(str)
    .str.split()
    .str.len()
)

# SPSS-kompatible Variablennamen
df = df.rename(columns={
    "price (USD)": "price_USD",
    "BERT discrete": "BERT_discrete",
    "BERT continuous": "BERT_continuous"
})

# Als SPSS-Datei speichern
pyreadstat.write_sav(df, "Analyse_mit_review_length.sav")

print("Fertig!")
```

```python
## Quadratisch gewichtetes Cohen's Kappa

from sklearn.metrics import cohen_kappa_score

# Suchgüter
df_search = df[df["good_type"] == "search"]

kappa_search = cohen_kappa_score(
    df_search["stars"],
    df_search["BERT discrete"],
    weights="quadratic"
)

# Erfahrungsgüter
df_exp = df[df["good_type"] == "experience"]

kappa_exp = cohen_kappa_score(
    df_exp["stars"],
    df_exp["BERT discrete"],
    weights="quadratic"
)

print("Quadratic weighted kappa")
print(f"Search:      {kappa_search:.4f}")
print(f"Experience: {kappa_exp:.4f}")
```
